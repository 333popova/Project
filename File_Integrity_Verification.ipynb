```python
import hashlib
def calculate_hash(filename):
  sha256=hashlib.sha256()
  with open(filename, "rb") as file:
    for block in iter(lambda: file.read(4096),b""):
      sha256.update(block)
      return sha256.hexdigest()

```

```python
with open ("test.txt","w")as file:
  file.write("This is good!")
```

```python
result=calculate_hash("test.txt")
print("SHA-256 Hash: ", result)
```

```python
def save_hash(filename):
  result= calculate_hash(filename)
  with open ("result_hash.txt",'w') as file:
    file.write(result)
    print("Hash saved successfully")

```

```python
save_hash("test.txt")
```

```python
def verify_file(filename):
  current_hash=calculate_hash(filename)
  with open ("result_hash.txt", "r")as file:
    result_hash=file.read()
    if current_hash==result_hash:
      print("Good - File is unchanged")
    else:
      print("Warning - File has been modified!!!")
```

```python
verify_file("test.txt")
```

```python
def main():
  while True:
    print(""" *******File Integrity Verification (How it works)
              1. Calculate file hash(SHA-256)
              2. Save original hash
              3. Verify the integrity
              4. Exit""")
    choice= input("Please choose one of the options: ")
    if choice=="1":
      filename=input("Enter file name: ")
      result=calculate_hash(filename)
      print("SHA-256 Hash: ", result)
    elif choice=="2":
        filename=input("Enter file name: ")
        save_hash(filename)
    elif choice=="3":
      filename=input("Enter file name: ")
      verify_file(filename)
    elif choice=="4":
      print("""*******Thank you!
       You are exit from File Integrity Verification""")
      break
    else:
      print("Sorry, you need choose options 1,2,3 or 4")





```
