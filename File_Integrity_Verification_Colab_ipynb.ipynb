```python
"""Use Case Scenario for Cryptography
15. File Integrity Verification:
Develop a tool to calculate and verify the hash values
of files, ensuring their integrity during transfer or storage"""

import hashlib
#import hashlib library to calculate sha256() hashes

def calculate_hash(filename):
# create a function to calculate  the file's SHA-256 hash
  sha256=hashlib.sha256()
  #create a SHA-256 hash object
  try:
# try to open and read  the file
    with open(filename, "rb") as file:
      for block in iter(lambda: file.read(4096),b""):
#read the file in blocks of 4096 bytes to avoid using too much memory(RAM).
#iter() repeats the reading operations until it finds
#an empty byte string (b""),
#which means the end of the file has been reached
#rb read the file in binary mode(allows SHA-256 calculation for any file type)
        sha256.update(block)
        #update the hash with each block
    return sha256.hexdigest()
    #return the calculated hash in hexadecimal format
  except FileNotFoundError:
    # handle the error if the file is not found
    print("Error: File not found!")
  except OSError:
    # handle errors when the file cannot be accessed
    print("Error: Unable to read or access the file.")

def save_hash(filename):
 # create a function to save the file's SHA-256 hash
  result = calculate_hash(filename)
  # stores the SHA-256 hash string returned by function calculate_hash
  if result is None:
    #check if function calculate_hash did not return a result
    return
  else:
     #if the hash was calculated successfully, try to save
     #the hash in separate file named filename.sha256
    try:
      hash_filename= filename+".sha256"
      with open (hash_filename,'w') as file:
        #w- opens the file in write mode,
        #with- automatically closes the file when finished
        file.write(result)
      print("Hash saved successfully")
    except OSError:
    # handle errors when the file cannot be accessed
      print("Error: Unable to save the hash file.")


def verify_file(filename):
  # create a function to verify whether the file has been modified
  current_hash=calculate_hash(filename)
   # stores the SHA-256 hash string returned by function calculate_hash

  if current_hash is None:
    #check if function calculate_hash did not return a result
    return
  else:
    #check if the hash calculation was successful
    # then try to open the hash in separate file named filename.sha256
    try:
      hash_filename= filename+".sha256"
      with open (hash_filename, "r")as file:
       result_hash=file.read().strip()
       # stores the saved hash
       # strip() removes unnecessary whitespace
       if current_hash==result_hash:
        # compare the hashes  to check whether they match
           print("Good - File is unchanged")
       else:
           print("Warning - File has been modified!!!")
    except FileNotFoundError:
     # handle the error if the file is not found
      print("Original hash not found. Please save the hash first")
    except OSError:
       # handle errors when the file cannot be accessed
      print("Error: Unable to read the original hash file.")


# - VERIFICATION HISTORY AND MULTIPLE FILES

import csv
import os
from datetime import datetime


def save_verification_history(filename, status):
    # Save verification results to a CSV file
    with open("verification_history.csv", "a", newline="") as file:
        writer = csv.writer(file)

        if file.tell() == 0:
            writer.writerow(["Filename", "Date", "Status"])

        writer.writerow([
            filename,
            datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            status
        ])


def view_verification_history():
    # Display all previous verification results
    if not os.path.exists("verification_history.csv"):
        print("No verification history available.")
        return

    with open("verification_history.csv", "r") as file:
        reader = csv.reader(file)

        print("\nVerification History")
        print("--------------------")

        for row in reader:
            print(" | ".join(row))


def verify_multiple_files():
    # Ask the user for multiple filenames
    filenames = input("Enter filenames separated by commas: ")

    file_list = filenames.split(",")

    print("\nMultiple File Verification")
    print("--------------------------")

    for filename in file_list:
        filename = filename.strip()

        if not filename:
            continue

        print("\nChecking:", filename)

        current_hash = calculate_hash(filename)

        if current_hash is None:
            status = "File not found or inaccessible"

        else:
            try:
                with open(filename + ".sha256", "r") as file:
                    original_hash = file.read().strip()

                if current_hash == original_hash:
                    status = "Unchanged"
                else:
                    status = "Modified"

            except FileNotFoundError:
                status = "Original hash missing"

            except OSError:
                status = "Unable to read original hash"

        print("Result:", status)

        # Automatically record the result
        save_verification_history(filename, status)

    print("\nAll files checked and history saved!")


def main():
  #creates the main function which controls the whole program
  while True:
  # create the loop that keeps the menu running until the user chooses to exit

    print(""" *******File Integrity Verification*******
        1. Calculate file hash (SHA-256)
        2. Save original hash
        3. Verify file integrity
        4. Verify multiple files
        5. View verification history
        6. Exit""")
    choice= input("Please choose one of the options: ")
    #ask the user to choose an option.
    if choice=="1":
      # user chooses option 1
      filename=input("Enter file name: ")
      result=calculate_hash(filename)
      #call the function to calculate  the file's SHA-256 hash and
      # stores file's hash SHA-256 in variable result
      if result is not None:
      #check that the hash calculation was successful

        print("SHA-256 Hash: ", result)
    elif choice=="2":
      # check whether the user choose 2
        filename=input("Enter file name: ")
        save_hash(filename)
    elif choice=="3":
      # check whether the user choose 3
      filename=input("Enter file name: ")
      verify_file(filename)
    elif choice=="4":
            # Verify multiple files
            verify_multiple_files()



    elif choice=="5":
            # View verification history
            view_verification_history()

    elif choice=="6":
       # Exit the application
        print("*******Thank you!")
        print("You are exiting File Integrity Verification")
        break

    else:
            print("Sorry, please choose an option from 1 to 6")

if __name__ == "__main__":
    main()
```
